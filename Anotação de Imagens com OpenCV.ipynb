## Leitura e exibição da imagem

Nesta primeira etapa, o objetivo é apenas **carregar e visualizar a imagem original** utilizando o OpenCV.

A imagem é enviada para o Google Colab e lida pela função `cv2.imread()`. Depois, utilizamos `cv2_imshow()` para exibi-la no notebook.

Também é importante lembrar que, no OpenCV, o ponto localizado no **canto superior esquerdo da imagem** corresponde à origem do sistema de coordenadas, ou seja:

$$
(x = 0,\ y = 0)
$$

O eixo **x** aumenta da esquerda para a direita, enquanto o eixo **y** aumenta de cima para baixo.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

if img is None:
    print("Não foi possível carregar a imagem.")
else:
    cv2_imshow(img)

    # Valor do pixel no ponto (0, 0)
    B, G, R = img[0, 0]

    print(f"(x=0, y=0) → R:{R} G:{G} B:{B}")
```

## Desenhando uma Linha na Imagem

Nesta etapa, utilizamos a função `cv2.line()` para desenhar uma linha sobre a imagem.

### Observações

* `img.copy()` cria uma cópia da imagem original.
* `pointA = (200, 80)` define o ponto inicial da linha.
* `pointB = (450, 80)` define o ponto final da linha.
* A linha possui **250 pixels de comprimento**.
* `(255, 255, 0)` define a cor no padrão BGR.
* `thickness=3` define a espessura da linha em 3 pixels.
* `cv2.line()` desenha a linha sobre a imagem.
* Para o Google Colab, `cv2.imshow()` foi substituído por `cv2_imshow()`.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageLine = img.copy()

# Pontos inicial e final
pointA = (200, 80)
pointB = (450, 80)

# Desenho da linha
cv2.line(
    imageLine,
    pointA,
    pointB,
    (255, 255, 0),
    thickness=3
)

# Exibição da imagem
cv2_imshow(imageLine)
```

## Desenhando um Círculo na Imagem

Nesta etapa, utilizamos a função `cv2.circle()` para **desenhar um círculo sobre a imagem**.

Primeiro, é criada uma cópia da imagem original. Em seguida, são definidas as coordenadas do **centro do círculo** e o seu **raio**.

Neste exemplo, o círculo possui centro em `(415, 190)`, raio de **100 pixels**, cor vermelha e espessura de **3 pixels**.

### Observações

* `img.copy()` cria uma cópia da imagem original.
* `circle_center` define a posição do centro do círculo.
* `radius` define o tamanho do círculo.
* `(0, 0, 255)` representa a cor **vermelha** no padrão BGR.
* `thickness=3` define a espessura da borda.
* `cv2.LINE_AA` deixa o contorno do círculo mais suave.
* `cv2.circle()` realiza o desenho do círculo na imagem.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageCircle = img.copy()

# Centro e raio do círculo
circle_center = (415, 190)
radius = 100

# Desenho do círculo
cv2.circle(
    imageCircle,
    circle_center,
    radius,
    (0, 0, 255),
    thickness=3,
    lineType=cv2.LINE_AA
)

# Exibição da imagem
cv2_imshow(imageCircle)
```

## Desenhando um Círculo Preenchido

Nesta etapa, utilizamos novamente a função `cv2.circle()`, mas agora para criar um **círculo totalmente preenchido**.

A principal diferença em relação ao exemplo anterior está no parâmetro `thickness`. Ao utilizar o valor `-1`, o OpenCV preenche todo o interior do círculo com a cor escolhida.

Neste exemplo, o círculo possui centro em `(415, 190)`, raio de **100 pixels** e preenchimento na cor azul.

### Observações

* `img.copy()` cria uma cópia da imagem original.
* `circle_center` define o centro do círculo.
* `radius = 100` define um raio de 100 pixels.
* `(255, 0, 0)` representa a cor **azul** no padrão BGR.
* `thickness=-1` faz com que o círculo fique completamente preenchido.
* `cv2.circle()` desenha o círculo sobre a imagem.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageFilledCircle = img.copy()

# Centro e raio do círculo
circle_center = (415, 190)
radius = 100

# Desenho do círculo preenchido
cv2.circle(
    imageFilledCircle,
    circle_center,
    radius,
    (255, 0, 0),
    thickness=-1,
    lineType=cv2.LINE_AA
)

# Exibição da imagem
cv2_imshow(imageFilledCircle)
```

## Desenhando um Retângulo na Imagem

Nesta etapa, utilizamos a função `cv2.rectangle()` para desenhar um retângulo sobre a imagem. Para criar o retângulo, são definidos dois pontos: o ponto inicial, que representa o canto superior esquerdo, e o ponto final, que representa o canto inferior direito. Neste exemplo, o retângulo é desenhado entre os pontos `(300, 115)` e `(475, 225)`, com a cor vermelha e espessura de 3 pixels.

### Observações

* `img.copy()` cria uma cópia da imagem original.
* `start_point` define o canto superior esquerdo do retângulo.
* `end_point` define o canto inferior direito.
* `(0, 0, 255)` representa a cor **vermelha** no padrão BGR.
* `thickness=3` define a espessura da borda em 3 pixels.
* `cv2.rectangle()` realiza o desenho do retângulo na imagem.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageRectangle = img.copy()

# Pontos inicial e final do retângulo
start_point = (300, 115)
end_point = (475, 225)

# Desenho do retângulo
cv2.rectangle(
    imageRectangle,
    start_point,
    end_point,
    (0, 0, 255),
    thickness=3,
    lineType=cv2.LINE_8
)

# Exibição da imagem
cv2_imshow(imageRectangle)
```

## Desenhando Elipses na Imagem

Nesta etapa, utilizamos a função `cv2.ellipse()` para desenhar elipses sobre a imagem. Diferente do círculo, a elipse possui dois eixos que determinam seu formato. Também é possível definir o **ângulo de rotação** da elipse e os ângulos inicial e final do desenho. Neste exemplo, são desenhadas duas elipses no mesmo ponto: uma azul na horizontal e outra vermelha na vertical.

### Observações

* `ellipse_center` define o centro das elipses.
* `axis1 = (100, 50)` define os eixos da elipse horizontal.
* `axis2 = (125, 50)` define os eixos da segunda elipse.
* O ângulo `0` mantém a primeira elipse na horizontal.
* O ângulo `90` rotaciona a segunda elipse, deixando-a na vertical.
* Os valores `0` e `360` fazem com que a elipse seja desenhada por completo.
* `(255, 0, 0)` representa a cor **azul** e `(0, 0, 255)` representa a cor **vermelha** no padrão BGR.
* `thickness=3` define a espessura do contorno.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageEllipse = img.copy()

# Centro das elipses
ellipse_center = (415, 190)

# Eixos das elipses
axis1 = (100, 50)
axis2 = (125, 50)

# Elipse horizontal azul
cv2.ellipse(
    imageEllipse,
    ellipse_center,
    axis1,
    0, 0, 360,
    (255, 0, 0),
    thickness=3
)

# Elipse vertical vermelha
cv2.ellipse(
    imageEllipse,
    ellipse_center,
    axis2,
    90, 0, 360,
    (0, 0, 255),
    thickness=3
)

# Exibição da imagem
cv2_imshow(imageEllipse)
```

## Desenhando uma Semi-Elipse

Nesta etapa, utilizamos novamente a função `cv2.ellipse()`, mas agora desenhando apenas metade das elipses.

A primeira elipse possui apenas o contorno azul, enquanto a segunda possui preenchimento vermelho. Para desenhar somente uma parte da elipse, são utilizados os ângulos inicial e final.

### Observações

* `ellipse_center` define o centro das elipses.
* `axis1 = (100, 50)` define o tamanho dos eixos.
* Os ângulos permitem controlar qual parte da elipse será desenhada.
* A primeira elipse possui apenas o contorno azul.
* A segunda elipse é vermelha e preenchida.
* `thickness=3` cria apenas o contorno da elipse.
* Um valor negativo em `thickness` faz com que a forma seja preenchida.
* Os valores de `0°` e `180°` permitem representar metade de uma elipse.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
halfEllipse = img.copy()

# Centro e eixos das elipses
ellipse_center = (415, 190)
axis1 = (100, 50)

# Semi-elipse azul com contorno
cv2.ellipse(
    halfEllipse,
    ellipse_center,
    axis1,
    0, 180, 360,
    (255, 0, 0),
    thickness=3
)

# Semi-elipse vermelha preenchida
cv2.ellipse(
    halfEllipse,
    ellipse_center,
    axis1,
    0, 0, 180,
    (0, 0, 255),
    thickness=-1
)

# Exibição da imagem
cv2_imshow(halfEllipse)
```

## Adicionando Texto na Imagem

Nesta etapa, utilizamos a função `cv2.putText()` para adicionar um texto sobre a imagem.

### Observações

* `text` define o texto que será escrito na imagem.
* `org = (50, 350)` define a posição onde o texto será colocado.
* `FONT_HERSHEY_COMPLEX` define o tipo da fonte.
* `fontScale = 1.5` define o tamanho do texto.
* `(250, 225, 100)` define a cor no padrão BGR.
* `cv2.putText()` adiciona o texto sobre a cópia da imagem original.
* Para o Google Colab, `cv2.imshow()` foi substituído por `cv2_imshow()`.


```python
import cv2
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Cópia da imagem original
imageText = img.copy()

# Texto e posição
text = 'I am a Happy dog!'
org = (50, 350)

# Adicionando o texto
cv2.putText(
    imageText,
    text,
    org,
    fontFace=cv2.FONT_HERSHEY_COMPLEX,
    fontScale=1.5,
    color=(250, 225, 100)
)

# Exibição da imagem
cv2_imshow(imageText)
```
