## Separação dos Canais BGR

Nesta etapa, as imagens **Outdoor** e **Indoor** são separadas nos canais B, G e R.

### Observações

* `cv2.split()` separa os três canais.
* Cada canal é mostrado em tons de cinza.
* A comparação permite observar o efeito da iluminação sobre cada canal.


```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload e leitura
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Divide a imagem em Outdoor e Indoor
height, width = img.shape[:2]
outdoor = img[:, :width // 2]
indoor = img[:, width // 2:]

# Separa os canais
B1, G1, R1 = cv2.split(outdoor)
B2, G2, R2 = cv2.split(indoor)

# Converte as imagens originais para RGB
outdoor_rgb = cv2.cvtColor(outdoor, cv2.COLOR_BGR2RGB)
indoor_rgb = cv2.cvtColor(indoor, cv2.COLOR_BGR2RGB)

# Exibição
fig, axes = plt.subplots(2, 4, figsize=(12, 6))

axes[0, 0].imshow(outdoor_rgb)
axes[0, 0].set_title("Outdoor")

axes[0, 1].imshow(B1, cmap="gray")
axes[0, 1].set_title("B")

axes[0, 2].imshow(G1, cmap="gray")
axes[0, 2].set_title("G")

axes[0, 3].imshow(R1, cmap="gray")
axes[0, 3].set_title("R")

axes[1, 0].imshow(indoor_rgb)
axes[1, 0].set_title("Indoor")

axes[1, 1].imshow(B2, cmap="gray")
axes[1, 1].set_title("B")

axes[1, 2].imshow(G2, cmap="gray")
axes[1, 2].set_title("G")

axes[1, 3].imshow(R2, cmap="gray")
axes[1, 3].set_title("R")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Espaço de Cores LAB

Nesta etapa, as imagens são convertidas do espaço de cores BGR para o espaço **LAB**.

### Observações

* O canal `L` representa a luminosidade.
* O canal `a` representa variações entre verde e magenta.
* O canal `b` representa variações entre azul e amarelo.
* A conversão é feita com `cv2.cvtColor()`.
* No LAB, a informação de brilho fica separada da informação de cor.


```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload e leitura
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Divide a imagem em Outdoor e Indoor
height, width = img.shape[:2]
bright = img[:, :width // 2]
dark = img[:, width // 2:]

# Conversão para LAB
brightLAB = cv2.cvtColor(bright, cv2.COLOR_BGR2LAB)
darkLAB = cv2.cvtColor(dark, cv2.COLOR_BGR2LAB)

# Separação dos canais
L1, A1, B1 = cv2.split(brightLAB)
L2, A2, B2 = cv2.split(darkLAB)

# Conversão das imagens originais para RGB
bright_rgb = cv2.cvtColor(bright, cv2.COLOR_BGR2RGB)
dark_rgb = cv2.cvtColor(dark, cv2.COLOR_BGR2RGB)

# Exibição
fig, axes = plt.subplots(2, 4, figsize=(12, 6))

axes[0, 0].imshow(bright_rgb)
axes[0, 0].set_title("Outdoor")

axes[0, 1].imshow(L1, cmap="gray")
axes[0, 1].set_title("L")

axes[0, 2].imshow(A1, cmap="gray")
axes[0, 2].set_title("a")

axes[0, 3].imshow(B1, cmap="gray")
axes[0, 3].set_title("b")

axes[1, 0].imshow(dark_rgb)
axes[1, 0].set_title("Indoor")

axes[1, 1].imshow(L2, cmap="gray")
axes[1, 1].set_title("L")

axes[1, 2].imshow(A2, cmap="gray")
axes[1, 2].set_title("a")

axes[1, 3].imshow(B2, cmap="gray")
axes[1, 3].set_title("b")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Espaço de Cores HSV

### Observações

* `H` representa a matiz (cor).
* `S` representa a saturação da cor.
* `V` representa a intensidade ou brilho.
* `cv2.cvtColor()` realiza a conversão para HSV.
* O HSV facilita a identificação e separação de cores.


```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload e leitura
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Divide a imagem em Outdoor e Indoor
height, width = img.shape[:2]
bright = img[:, :width // 2]
dark = img[:, width // 2:]

# Conversão para HSV
brightHSV = cv2.cvtColor(bright, cv2.COLOR_BGR2HSV)
darkHSV = cv2.cvtColor(dark, cv2.COLOR_BGR2HSV)

# Separação dos canais
H1, S1, V1 = cv2.split(brightHSV)
H2, S2, V2 = cv2.split(darkHSV)

# Imagens originais em RGB
bright_rgb = cv2.cvtColor(bright, cv2.COLOR_BGR2RGB)
dark_rgb = cv2.cvtColor(dark, cv2.COLOR_BGR2RGB)

# Exibição
fig, axes = plt.subplots(2, 4, figsize=(12, 6))

axes[0, 0].imshow(bright_rgb)
axes[0, 0].set_title("Outdoor")
axes[0, 1].imshow(H1, cmap="gray")
axes[0, 1].set_title("H")
axes[0, 2].imshow(S1, cmap="gray")
axes[0, 2].set_title("S")
axes[0, 3].imshow(V1, cmap="gray")
axes[0, 3].set_title("V")

axes[1, 0].imshow(dark_rgb)
axes[1, 0].set_title("Indoor")
axes[1, 1].imshow(H2, cmap="gray")
axes[1, 1].set_title("H")
axes[1, 2].imshow(S2, cmap="gray")
axes[1, 2].set_title("S")
axes[1, 3].imshow(V2, cmap="gray")
axes[1, 3].set_title("V")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Aplicando um Limiar para Segmentação

Nesta etapa, usamos os valores de uma cor verde como referência para encontrar pixels semelhantes na imagem.

### Observações

* Foi definido um intervalo de `±40` em torno do valor da cor verde.
* `cv2.inRange()` cria uma máscara com os pixels que estão dentro desse intervalo.
* `cv2.bitwise_and()` mantém apenas os pixels selecionados pela máscara.
* O mesmo processo é aplicado nos espaços **BGR, HSV, YCrCb e LAB**.
* Os resultados permitem comparar qual espaço de cores segmenta melhor a região verde.


```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from google.colab import files

# Upload e leitura da imagem
uploaded = files.upload()
filename = next(iter(uploaded))
bright = cv2.imread(filename)

# Conversões
brightHSV = cv2.cvtColor(bright, cv2.COLOR_BGR2HSV)
brightYCB = cv2.cvtColor(bright, cv2.COLOR_BGR2YCrCb)
brightLAB = cv2.cvtColor(bright, cv2.COLOR_BGR2LAB)

# Cor verde de referência
bgr = np.array([40, 158, 16], dtype=np.int16)
thresh = 40

# BGR
minBGR = np.clip(bgr - thresh, 0, 255).astype(np.uint8)
maxBGR = np.clip(bgr + thresh, 0, 255).astype(np.uint8)

maskBGR = cv2.inRange(bright, minBGR, maxBGR)
resultBGR = cv2.bitwise_and(bright, bright, mask=maskBGR)

# HSV
hsv = cv2.cvtColor(
    np.uint8([[bgr.astype(np.uint8)]]),
    cv2.COLOR_BGR2HSV
)[0][0].astype(np.int16)

minHSV = np.clip(hsv - thresh, 0, 255).astype(np.uint8)
maxHSV = np.clip(hsv + thresh, 0, 255).astype(np.uint8)

maskHSV = cv2.inRange(brightHSV, minHSV, maxHSV)
resultHSV = cv2.bitwise_and(brightHSV, brightHSV, mask=maskHSV)

# YCrCb
ycb = cv2.cvtColor(
    np.uint8([[bgr.astype(np.uint8)]]),
    cv2.COLOR_BGR2YCrCb
)[0][0].astype(np.int16)

minYCB = np.clip(ycb - thresh, 0, 255).astype(np.uint8)
maxYCB = np.clip(ycb + thresh, 0, 255).astype(np.uint8)

maskYCB = cv2.inRange(brightYCB, minYCB, maxYCB)
resultYCB = cv2.bitwise_and(brightYCB, brightYCB, mask=maskYCB)

# LAB
lab = cv2.cvtColor(
    np.uint8([[bgr.astype(np.uint8)]]),
    cv2.COLOR_BGR2LAB
)[0][0].astype(np.int16)

minLAB = np.clip(lab - thresh, 0, 255).astype(np.uint8)
maxLAB = np.clip(lab + thresh, 0, 255).astype(np.uint8)

maskLAB = cv2.inRange(brightLAB, minLAB, maxLAB)
resultLAB = cv2.bitwise_and(brightLAB, brightLAB, mask=maskLAB)

# Exibição
print("Resultado BGR:")
cv2_imshow(resultBGR)

print("Resultado HSV:")
cv2_imshow(resultHSV)

print("Resultado YCrCb:")
cv2_imshow(resultYCB)

print("Resultado LAB:")
cv2_imshow(resultLAB)
```

## Teste do Mesmo Limiar em Diferentes Iluminações
## Observações
O mesmo valor de limiar ±40 foi aplicado em uma imagem clara e em uma imagem escura.
A segmentação foi testada nos espaços BGR, HSV, YCrCb e LAB.
Os resultados mostram que mudanças de iluminação podem fazer o mesmo limiar funcionar em uma imagem e falhar em outra.
Isso mostra que escolher o limite apenas por tentativa e erro não é uma solução confiável.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload da imagem com os dois cubos
uploaded = files.upload()
filename = next(iter(uploaded))

img = cv2.imread(filename)

# Divide a imagem em Outdoor e Indoor
height, width = img.shape[:2]

bright = img[:, :width // 2]
dark = img[:, width // 2:]

# Cor verde de referência
bgr = np.array([40, 158, 16], dtype=np.int16)
thresh = 40


def segment_green(image):

    # Conversões
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    ycrcb = cv2.cvtColor(image, cv2.COLOR_BGR2YCrCb)
    lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

    # BGR
    minBGR = np.clip(bgr - thresh, 0, 255).astype(np.uint8)
    maxBGR = np.clip(bgr + thresh, 0, 255).astype(np.uint8)

    maskBGR = cv2.inRange(image, minBGR, maxBGR)

    # HSV
    hsv_ref = cv2.cvtColor(
        np.uint8([[bgr.astype(np.uint8)]]),
        cv2.COLOR_BGR2HSV
    )[0][0].astype(np.int16)

    minHSV = np.clip(hsv_ref - thresh, 0, 255).astype(np.uint8)
    maxHSV = np.clip(hsv_ref + thresh, 0, 255).astype(np.uint8)

    maskHSV = cv2.inRange(hsv, minHSV, maxHSV)

    # YCrCb
    ycb_ref = cv2.cvtColor(
        np.uint8([[bgr.astype(np.uint8)]]),
        cv2.COLOR_BGR2YCrCb
    )[0][0].astype(np.int16)

    minYCB = np.clip(ycb_ref - thresh, 0, 255).astype(np.uint8)
    maxYCB = np.clip(ycb_ref + thresh, 0, 255).astype(np.uint8)

    maskYCB = cv2.inRange(ycrcb, minYCB, maxYCB)

    # LAB
    lab_ref = cv2.cvtColor(
        np.uint8([[bgr.astype(np.uint8)]]),
        cv2.COLOR_BGR2LAB
    )[0][0].astype(np.int16)

    minLAB = np.clip(lab_ref - thresh, 0, 255).astype(np.uint8)
    maxLAB = np.clip(lab_ref + thresh, 0, 255).astype(np.uint8)

    maskLAB = cv2.inRange(lab, minLAB, maxLAB)

    # Aplica as máscaras na imagem original
    resultBGR = cv2.bitwise_and(image, image, mask=maskBGR)
    resultHSV = cv2.bitwise_and(image, image, mask=maskHSV)
    resultYCB = cv2.bitwise_and(image, image, mask=maskYCB)
    resultLAB = cv2.bitwise_and(image, image, mask=maskLAB)

    return resultBGR, resultHSV, resultYCB, resultLAB


# Segmentação
bright_results = segment_green(bright)
dark_results = segment_green(dark)

# Exibição
fig, axes = plt.subplots(2, 4, figsize=(14, 7))

titles = ["BGR", "HSV", "YCrCb", "LAB"]

for i in range(4):

    axes[0, i].imshow(
        cv2.cvtColor(bright_results[i], cv2.COLOR_BGR2RGB)
    )
    axes[0, i].set_title(titles[i])
    axes[0, i].axis("off")

    axes[1, i].imshow(
        cv2.cvtColor(dark_results[i], cv2.COLOR_BGR2RGB)
    )
    axes[1, i].set_title(titles[i])
    axes[1, i].axis("off")

axes[0, 0].set_ylabel("Outdoor")
axes[1, 0].set_ylabel("Indoor")

plt.tight_layout()
plt.show()
```

## Segmentação da Cor Amarela

Nesta etapa, aplicamos limites de cor para identificar as regiões amarelas nas imagens **Outdoor** e **Indoor**.

### Observações

* A segmentação foi realizada nos espaços **BGR, HSV, YCrCb e LAB**.
* `Min` e `Max` definem a faixa de valores aceita para cada espaço de cor.
* `cv2.inRange()` identifica os pixels dentro desses limites.
* O mesmo limite é aplicado nas duas condições de iluminação.
* Os resultados mostram que a mudança de iluminação pode afetar bastante a detecção da cor.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload da imagem com os dois cubos
uploaded = files.upload()
filename = next(iter(uploaded))
img = cv2.imread(filename)

# Separação Outdoor e Indoor
height, width = img.shape[:2]
bright = img[:, :width // 2]
dark = img[:, width // 2:]

# Limites para cada espaço de cor
minBGR = np.array([0, 162, 186])
maxBGR = np.array([46, 242, 255])

minHSV = np.array([0, 208, 186])
maxHSV = np.array([47, 255, 255])

minYCB = np.array([0, 147, 116])
maxYCB = np.array([66, 227, 196])

minLAB = np.array([167, 80, 169])
maxLAB = np.array([247, 161, 249])


def segmentar(image):

    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    ycb = cv2.cvtColor(image, cv2.COLOR_BGR2YCrCb)
    lab = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

    maskBGR = cv2.inRange(image, minBGR, maxBGR)
    maskHSV = cv2.inRange(hsv, minHSV, maxHSV)
    maskYCB = cv2.inRange(ycb, minYCB, maxYCB)
    maskLAB = cv2.inRange(lab, minLAB, maxLAB)

    resultBGR = cv2.bitwise_and(image, image, mask=maskBGR)
    resultHSV = cv2.bitwise_and(image, image, mask=maskHSV)
    resultYCB = cv2.bitwise_and(image, image, mask=maskYCB)
    resultLAB = cv2.bitwise_and(image, image, mask=maskLAB)

    return [resultBGR, resultHSV, resultYCB, resultLAB]


def mostrar_resultado(image, titulo):

    results = segmentar(image)

    nomes = ["BGR", "HSV", "YCrCb", "LAB"]

    mins = [minBGR, minHSV, minYCB, minLAB]
    maxs = [maxBGR, maxHSV, maxYCB, maxLAB]

    fig, axes = plt.subplots(1, 4, figsize=(14, 4))

    for i in range(4):

        axes[i].imshow(
            cv2.cvtColor(results[i], cv2.COLOR_BGR2RGB)
        )

        axes[i].set_title(nomes[i])
        axes[i].axis("off")

        axes[i].text(
            0.5, -0.10,
            f"Min {mins[i]}\nMax {maxs[i]}",
            transform=axes[i].transAxes,
            ha="center",
            fontsize=10
        )

    plt.suptitle(titulo)
    plt.tight_layout()
    plt.show()


# Resultado Outdoor
mostrar_resultado(bright, "Outdoor")

# Resultado Indoor
mostrar_resultado(dark, "Indoor")
```

* `plt.hist2d()` cria um histograma 2D mostrando a distribuição dos valores.
* O gráfico ajuda a identificar a faixa de valores mais comum para uma determinada cor.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from google.colab import files

# Upload de várias imagens da mesma cor
uploaded = files.upload()

# Arrays para armazenar os canais
B, G, R = [], [], []
H, S = [], []
Cr, Cb = [], []
A, LabB = [], []

for filename in uploaded.keys():

    img = cv2.imread(filename)

    # BGR
    b, g, r = cv2.split(img)

    B.extend(b.ravel())
    G.extend(g.ravel())
    R.extend(r.ravel())

    # HSV
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    h, s, v = cv2.split(hsv)

    H.extend(h.ravel())
    S.extend(s.ravel())

    # YCrCb
    ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
    y, cr, cb = cv2.split(ycrcb)

    Cr.extend(cr.ravel())
    Cb.extend(cb.ravel())

    # LAB
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, lab_b = cv2.split(lab)

    A.extend(a.ravel())
    LabB.extend(lab_b.ravel())

# Conversão para arrays NumPy
B = np.array(B)
G = np.array(G)
R = np.array(R)

H = np.array(H)
S = np.array(S)

Cr = np.array(Cr)
Cb = np.array(Cb)

A = np.array(A)
LabB = np.array(LabB)

# Criação dos histogramas
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

# RGB - B x G
hist = axes[0, 0].hist2d(
    B, G,
    bins=30,
    range=[[0, 255], [0, 255]],
    norm=LogNorm()
)
axes[0, 0].set_title("RGB")
axes[0, 0].set_xlabel("B")
axes[0, 0].set_ylabel("G")
fig.colorbar(hist[3], ax=axes[0, 0])

# RGB - B x R
hist = axes[0, 1].hist2d(
    B, R,
    bins=30,
    range=[[0, 255], [0, 255]],
    norm=LogNorm()
)
axes[0, 1].set_title("RGB")
axes[0, 1].set_xlabel("B")
axes[0, 1].set_ylabel("R")
fig.colorbar(hist[3], ax=axes[0, 1])

# RGB - R x G
hist = axes[0, 2].hist2d(
    R, G,
    bins=30,
    range=[[0, 255], [0, 255]],
    norm=LogNorm()
)
axes[0, 2].set_title("RGB")
axes[0, 2].set_xlabel("R")
axes[0, 2].set_ylabel("G")
fig.colorbar(hist[3], ax=axes[0, 2])

# HSV - H x S
hist = axes[1, 0].hist2d(
    H, S,
    bins=30,
    range=[[0, 179], [0, 255]],
    norm=LogNorm()
)
axes[1, 0].set_title("HSV")
axes[1, 0].set_xlabel("H")
axes[1, 0].set_ylabel("S")
fig.colorbar(hist[3], ax=axes[1, 0])

# YCrCb - Cr x Cb
hist = axes[1, 1].hist2d(
    Cr, Cb,
    bins=30,
    range=[[0, 255], [0, 255]],
    norm=LogNorm()
)
axes[1, 1].set_title("YCrCb")
axes[1, 1].set_xlabel("Cr")
axes[1, 1].set_ylabel("Cb")
fig.colorbar(hist[3], ax=axes[1, 1])

# LAB - A x B
hist = axes[1, 2].hist2d(
    A, LabB,
    bins=30,
    range=[[0, 255], [0, 255]],
    norm=LogNorm()
)
axes[1, 2].set_title("LAB")
axes[1, 2].set_xlabel("A")
axes[1, 2].set_ylabel("B")
fig.colorbar(hist[3], ax=axes[1, 2])

plt.tight_layout()
plt.show()
```
