# Leitura de imagens com OpenCV

As operações básicas de processamento de imagens no OpenCV incluem **ler, exibir e salvar imagens**.

* `cv2.imread()` — carrega uma imagem para o programa.
* `cv2.imshow()` — exibe a imagem em uma janela.
* `cv2.imwrite()` — salva a imagem em um arquivo.

A função `cv2.imread()` permite definir diferentes modos de leitura:

* `cv2.IMREAD_COLOR` — carrega a imagem colorida.
* `cv2.IMREAD_GRAYSCALE` — carrega a imagem em tons de cinza.
* `cv2.IMREAD_UNCHANGED` — carrega a imagem

### Observação

O OpenCV permite escolher como a imagem será lida por meio do segundo parâmetro da função `cv2.imread()`. Essa escolha pode ser feita utilizando nomes das constantes ou valores numéricos:

* `1` (`cv2.IMREAD_COLOR`) → lê a imagem **colorida**, utilizando os canais **BGR**.
* `0` (`cv2.IMREAD_GRAYSCALE`) → lê a imagem em **tons de cinza**.
* `-1` (`cv2.IMREAD_UNCHANGED`) → mantém os **canais originais** da imagem.

Assim, os valores `1`, `0` e `-1` são formas mais simples de indicar o modo de leitura desejado.

````python
import cv2
from google.colab.patches import cv2_imshow

# Read an image

img_color = cv2.imread('test.jpg', cv2.IMREAD_COLOR)

img_grayscale = cv2.imread('test.jpg', cv2.IMREAD_GRAYSCALE)

img_unchanged = cv2.imread('test.jpg', cv2.IMREAD_UNCHANGED)

# Display the images

print("Imagem colorida:")
cv2_imshow(img_color)

print("Imagem em tons de cinza:")
cv2_imshow(img_grayscale)

print("Imagem com IMREAD_UNCHANGED:")
cv2_imshow(img_unchanged)
````

## **Leitura e escrita de imagens**

A função `cv2.imwrite()` é utilizada para **salvar uma imagem em um arquivo**. Depois de realizar a leitura e o processamento da imagem, podemos utilizar essa função para armazenar o resultado em um novo arquivo.

Sua sintaxe é:

```python
cv2.imwrite(filename, image)
```

* **`filename`** → nome do arquivo, incluindo sua extensão, como `.jpg` ou `.png`.
* **`image`** → variável que contém a imagem que será salva.

A função retorna `True` quando a imagem é salva com sucesso.

No exemplo abaixo, a imagem armazenada na variável `img_grayscale` é salva como `grayscale.jpg`:

```python
cv2.imwrite('grayscale.jpg', img_grayscale)
```

No Google Colab, o arquivo é salvo no ambiente de execução do notebook e pode ser acessado pela área de arquivos da sessão.



````python
# Importar a biblioteca OpenCV
import cv2

# Importar a função para exibir imagens no Google Colab
from google.colab.patches import cv2_imshow


# Modos de leitura disponíveis
READ_MODES = {
    "color": cv2.IMREAD_COLOR,
    "grayscale": cv2.IMREAD_GRAYSCALE,
    "unchanged": cv2.IMREAD_UNCHANGED,
}


# Função para ler uma imagem
def read_image(path, mode="color"):

    # Verifica se o modo escolhido é válido
    if mode not in READ_MODES:
        raise ValueError("unsupported read mode")

    # Lê a imagem utilizando o modo escolhido
    image = cv2.imread(str(path), READ_MODES[mode])

    # Verifica se a imagem foi carregada corretamente
    if image is None:
        raise FileNotFoundError(f"Could not decode image: {path}")

    return image


# Ler a imagem nos diferentes modos
img_color = read_image('test.jpg', "color")
img_grayscale = read_image('test.jpg', "grayscale")
img_unchanged = read_image('test.jpg', "unchanged")


# Exibir as imagens
print("Imagem colorida:")
cv2_imshow(img_color)

print("Imagem em tons de cinza:")
cv2_imshow(img_grayscale)

print("Imagem com os canais originais:")
cv2_imshow(img_unchanged)


# Salvar a imagem em tons de cinza
cv2.imwrite('grayscale.jpg', img_grayscale)
````

### O que foi adaptado para o Colab?

````python
import cv2
from pathlib import Path
from google.colab.patches import cv2_imshow

# Carregar a imagem
color = cv2.imread("/content/test.jpg")

# Verificar se a imagem foi carregada
if color is None:
    raise FileNotFoundError("A imagem não foi encontrada.")

# Salvar como PNG com compressão
cv2.imwrite(
    "/content/lossless-copy.png",
    color,
    [cv2.IMWRITE_PNG_COMPRESSION, 6]
)

# Salvar como JPEG com qualidade 90
cv2.imwrite(
    "/content/quality-90.jpg",
    color,
    [cv2.IMWRITE_JPEG_QUALITY, 90]
)

# Exibir a imagem
cv2_imshow(color)
````

### Verificação das imagens salvas

Depois de salvar as imagens em PNG e JPEG, carreguei os arquivos novamente para verificar se foram gravados corretamente.

No caso do PNG, comparei a imagem original com a imagem salva usando `np.array_equal()`. Como o PNG utiliza compressão sem perdas, os valores dos pixels devem permanecer exatamente iguais.

Já no JPEG, não é esperado que os pixels sejam exatamente iguais, pois esse formato utiliza compressão com perdas. Por isso, calculei o erro médio absoluto (`MAE`) para verificar a diferença entre a imagem original e a imagem JPEG.

Também converti os valores para `int16` antes de fazer a subtração, evitando problemas que poderiam ocorrer com o tipo `uint8`.

Dessa forma, foi possível confirmar que o PNG manteve os pixels originais, enquanto o JPEG apresentou pequenas diferenças devido à sua compressão.


````python
import cv2
import numpy as np

# Caminhos das imagens
png_path = "/content/lossless-copy.png"
jpeg_path = "/content/quality-90.jpg"

# Ler novamente as imagens que foram salvas
png_roundtrip = cv2.imread(png_path, cv2.IMREAD_COLOR)
jpeg_roundtrip = cv2.imread(jpeg_path, cv2.IMREAD_COLOR)

# Verificar se as imagens foram carregadas corretamente
if png_roundtrip is None:
    raise FileNotFoundError("Não foi possível ler o arquivo PNG.")

if jpeg_roundtrip is None:
    raise FileNotFoundError("Não foi possível ler o arquivo JPEG.")

# Verificar se o PNG manteve exatamente os mesmos pixels
if not np.array_equal(color, png_roundtrip):
    raise AssertionError("O PNG alterou os valores dos pixels.")

print("PNG: os pixels foram preservados corretamente.")

# Calcular o erro médio absoluto do JPEG
jpeg_mae = np.mean(
    np.abs(
        color.astype(np.int16) -
        jpeg_roundtrip.astype(np.int16)
    )
)

print(f"JPEG - erro médio absoluto: {jpeg_mae:.3f}")
````
