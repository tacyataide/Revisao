```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

nome = list(uploaded.keys())[0]
img = cv2.imread(nome)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

Kernel de Identidade

O Kernel de Identidade mantém a imagem sem alterações. Ele possui o valor 1 no centro e 0 nas demais posições.

A função cv2.filter2D() aplica o kernel sobre a imagem.

```python
kernel = np.array([[0, 0, 0],
                   [0, 1, 0],
                   [0, 0, 0]])

identity = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(identity, cv2.COLOR_BGR2RGB))
plt.title('Identidade')
plt.axis('off')

plt.show()
```

Desfoque usando Kernels

O desfoque (Blur) suaviza a imagem utilizando um kernel. Neste exemplo, usamos uma matriz 3×3 preenchida com 1 e dividida por 9. Quanto maior o kernel, maior será o desfoque aplicado.

```python
kernel = np.ones((3, 3), np.float32) / 9

blur = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title('Desfoque')
plt.axis('off')

plt.show()
```

Desfoque com cv2.blur()

O OpenCV possui a função cv2.blur(), que permite aplicar desfoque diretamente na imagem. O tamanho do kernel define a intensidade da suavização.

```python
img_blur = cv2.blur(img, (5, 5))

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(img_blur, cv2.COLOR_BGR2RGB))
plt.title('Desfocada')
plt.axis('off')

plt.show()
```

Desfoque Gaussiano

O Desfoque Gaussiano suaviza a imagem e reduz detalhes e ruídos. A função cv2.GaussianBlur() utiliza um kernel Gaussiano para realizar o desfoque.

```python
blur = cv2.GaussianBlur(img, (5, 5), 0)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title('Desfoque Gaussiano')
plt.axis('off')

plt.show()
```

Desfoque pela Mediana

O Desfoque pela Mediana substitui cada pixel pelo valor mediano dos pixels da sua vizinhança. Esse método é útil para suavizar a imagem e reduzir ruídos.

```python
median = cv2.medianBlur(img, 5)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title('Desfoque pela Mediana')
plt.axis('off')

plt.show()
```

Nitidez usando Kernel

O Sharpen é utilizado para aumentar a nitidez da imagem, destacando detalhes e bordas. Para isso, aplicamos um kernel que intensifica o pixel central em relação aos pixels vizinhos.

```python
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

sharpen = cv2.filter2D(img, -1, kernel)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(sharpen, cv2.COLOR_BGR2RGB))
plt.title('Nitidez')
plt.axis('off')

plt.show()
```

Filtragem Bilateral

A Filtragem Bilateral é utilizada para reduzir ruídos da imagem enquanto preserva as bordas. Diferente de outros métodos de desfoque, ela considera tanto a distância entre os pixels quanto a diferença de cor entre eles.

```python
bf = cv2.bilateralFilter(img, 9, 75, 75)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Original')
plt.axis('off')

plt.subplot(122)
plt.imshow(cv2.cvtColor(bf, cv2.COLOR_BGR2RGB))
plt.title('Filtragem Bilateral')
plt.axis('off')

plt.show()
```
